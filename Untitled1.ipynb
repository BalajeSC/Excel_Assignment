The error `TypeError: 'int' object is not iterable` occurs because you're trying to pass an integer `i` directly to the `list()` constructor. The `list()` constructor expects an iterable (like a string, tuple, or another range object).

To fix this, you simply need to print the variable `i` directly, as `i` already holds the integer value you want.

```python
# Print 0 through 5
for i in range(6):
  print(i)

# Print 2 through 5
for i in range(2, 6):
  print(i)
```
