The error `TypeError: 'int' object is not iterable` occurs because you're trying to pass an integer `i` directly to the `list()` constructor. The `list()` constructor expects an iterable (like a string, tuple, or another range object).

To fix this, you simply need to print the variable `i` directly, as `i` already holds the integer value you want.

```python
# Print 0 through 5
for i in range(6):
  print(i)

# Print 2 through 5
for i in range(2, 6):
  print(i)
```

### Using `extend()` to add elements from another iterable

The `extend()` method appends all the items from an iterable (e.g., another list) to the end of the current list. It modifies the list in place and does not return a new list.

```python
list1 = [1, 2, 3]
list2 = [4, 5, 6]

print(f"Original list1: {list1}")
print(f"List to extend with (list2): {list2}")

list1.extend(list2)

print(f"List1 after extend(): {list1}")

# You can also extend with other iterables like tuples or strings
list3 = ['a', 'b']
tuple1 = ('c', 'd')

print(f"\nOriginal list3: {list3}")
print(f"Tuple to extend with (tuple1): {tuple1}")

list3.extend(tuple1)

print(f"List3 after extend() with a tuple: {list3}")

string1 = "ef"

print(f"\nOriginal list3: {list3}")
print(f"String to extend with (string1): {string1}")

list3.extend(string1)

print(f"List3 after extend() with a string: {list3}")
```
